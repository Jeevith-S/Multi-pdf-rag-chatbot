```python
all_documents = []

for filename in uploaded_files.keys():

    reader = PdfReader(filename)

    text = ""

    for page in reader.pages:

        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n"

    all_documents.append({
        "source": filename,
        "text": text
    })

print(f"Loaded {len(all_documents)} PDFs")
```

```python
for doc in all_documents:

    print("=" * 50)
    print("File:", doc["source"])
    print("=" * 50)

    print(doc["text"][:1000])  # first 1000 characters
    print("\n")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = text_splitter.split_text(text)

print(chunks)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "nomic-ai/nomic-embed-text-v1.5"
)
```

```python
tokenized_chunks = []

# Assuming all chunks belong to the first document uploaded
# and that 'all_documents' is populated from the uploaded file
source_filename = all_documents[0]["source"] if all_documents else "unknown_source"

for chunk_text in chunks: # Correctly iterate over the 'chunks' list

    tokens = tokenizer.tokenize(
        chunk_text # 'chunk_text' is the string content of the chunk
    )

    token_ids = tokenizer.encode(
        chunk_text,
        add_special_tokens=True
    )

    tokenized_chunks.append({
        "source": source_filename, # Use the common source for all chunks
        "text": chunk_text, # Use chunk_text as the text for this chunk
        "tokens": tokens,
        "token_ids": token_ids
    })
```

```python
for chunk in tokenized_chunks[:3]:

    print("Source:", chunk["source"])
    print("Text:", chunk["text"])
    print("Token IDs:", chunk["token_ids"])
    print("-" * 50)
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "nomic-ai/nomic-embed-text-v1.5",
    trust_remote_code=True
)
```

```python
def create_chunk_embeddings(all_chunks):

    texts = [
        chunk["text"]
        for chunk in all_chunks
    ]

    embeddings = embedding_model.encode(
        texts,
        convert_to_numpy=True
    )

    return embeddings
```

```python
chunk_embeddings = create_chunk_embeddings(
    tokenized_chunks
)
```

```python
import chromadb
```

```python
client = chromadb.PersistentClient(
    path="./chroma_db"
)
```

```python
collection = client.get_or_create_collection(
    name="pdf_documents"
)
```

```python
def store_in_vectordb(
    all_chunks,
    chunk_embeddings,
    collection
):

    collection.add(

        ids=[
            str(i)
            for i in range(len(all_chunks))
        ],

        documents=[
            chunk["text"]
            for chunk in all_chunks
        ],

        embeddings=
            chunk_embeddings.tolist(),

        metadatas=[
            {
                "source": chunk["source"]
            }
            for chunk in all_chunks
        ]
    )

    print(
        f"{len(all_chunks)} chunks stored successfully"
    )
```

```python
store_in_vectordb(
    tokenized_chunks,
    chunk_embeddings,
    collection
)
```

```python
def generate_answer(question, results):

    context = "\n\n".join(
        results["documents"][0]
    )

    prompt = f"""
    Use the provided context to answer the question.

    Context:
    {context}

    Question:
    {question}
    """

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",

        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```

```python
answer = generate_answer(
    question,
    results
)

print(answer)
```
